```python
import heapq
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Grafo
grafo = {
    'A': {'B': 4, 'C': 6},
    'B': {'D': 5, 'E': 9},
    'C': {'D': 3},
    'D': {'E': 2, 'F': 7},
    'E': {'F': 3},
    'F': {}
}

# Heurística
heuristica = {
    'A': 10,
    'B': 8,
    'C': 7,
    'D': 4,
    'E': 2,
    'F': 0
}

# Nombres de lugares
lugares = {
    'A': 'A\nAlcaldía',
    'B': 'B\nParque',
    'C': 'C\nMalecón',
    'D': 'D\nBellavista',
    'E': 'E\nKennedy',
    'F': 'F\nUniversidad\ndel Pacífico'
}

# Crear grafo visual
G = nx.DiGraph()

for nodo in grafo:
    for vecino, costo in grafo[nodo].items():
        G.add_edge(nodo, vecino, weight=costo)

pos = nx.spring_layout(G, k=5, seed=20)

# Algoritmo A*
def a_star(inicio, objetivo):

    cola = []
    heapq.heappush(cola, (0, inicio))

    costos = {inicio: 0}
    padres = {inicio: None}

    while cola:

        _, actual = heapq.heappop(cola)

        if actual == objetivo:
            break

        for vecino, costo in grafo[actual].items():

            nuevo_costo = costos[actual] + costo

            if vecino not in costos or nuevo_costo < costos[vecino]:

                costos[vecino] = nuevo_costo
                prioridad = nuevo_costo + heuristica[vecino]

                heapq.heappush(cola, (prioridad, vecino))
                padres[vecino] = actual

    ruta = []
    nodo = objetivo

    while nodo:
        ruta.append(nodo)
        nodo = padres[nodo]

    ruta.reverse()

    return ruta, costos[objetivo]


ruta_final, costo_total = a_star('A','F')

print("Ruta óptima encontrada:")
print(" → ".join(ruta_final))
print("Costo total:", costo_total)

fig, ax = plt.subplots()

def actualizar(frame):

    ax.clear()

    # nodos que se van activando
    nodos_activos = (frame // 2) + 1

    colores = []

    for i, nodo in enumerate(G.nodes()):

        if nodo in ruta_final[:nodos_activos]:
            colores.append("green")
        else:
            colores.append("lightgray")

    nx.draw(
        G,
        pos,
        labels=lugares,
        node_color=colores,
        node_size=2500,
        font_size=10,
        ax=ax
    )

    # pesos
    edge_labels = nx.get_edge_attributes(G, "weight")
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, ax=ax)

    # aristas activas
    aristas_ruta = list(zip(ruta_final[:-1], ruta_final[1:]))

    aristas_activas = frame // 2

    nx.draw_networkx_edges(
        G,
        pos,
        edgelist=aristas_ruta[:aristas_activas],
        edge_color="red",
        width=4,
        ax=ax
    )

    ax.set_title("Recorrido De La Ruta Más Corta Con El Algoritmo A* \n Para Llegar A la Universidad Del Pacífico")

plt.close()

anim = FuncAnimation(
    fig,
    actualizar,
    frames=len(ruta_final)*2,
    interval=1200
)

HTML(anim.to_html5_video())
```
