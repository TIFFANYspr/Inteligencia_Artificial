```python
import numpy as np

estrategias = ["Mantener precio", "Bajar precio"]

# Matriz de pagos (D1, Ara)
matriz = [
    [(8,8), (3,10)],
    [(10,3), (5,5)]
]

print("Matriz de pagos (D1, Ara):\n")

for i in range(2):
    for j in range(2):
        print(f"D1: {estrategias[i]} | Ara: {estrategias[j]} -> {matriz[i][j]}")

print("\nResultados posibles del juego:\n")

for i in range(2):
    for j in range(2):
        d1, ara = matriz[i][j]

        print(f"Estrategia elegida:")
        print(f"D1 -> {estrategias[i]}")
        print(f"Ara -> {estrategias[j]}")
        print(f"Ganancia D1 = {d1}")
        print(f"Ganancia Ara = {ara}")
        print("-------------------------")

print("\nSe puede concluir que:")
print("El equilibrio del juego ocurre cuando ambos supermercados bajan el precio.")
print("Resultado: (5,5)")
```
