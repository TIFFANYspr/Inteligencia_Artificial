```python
# Rutas
rutas = {
    "Buenaventura - Cali - Armenia": {
        "distancia": 210,
        "trafico": 6,
        "peajes": 3
    },

    "Buenaventura - Buga - Tulua - Armenia": {
        "distancia": 220,
        "trafico": 4,
        "peajes": 2
    },

    "Buenaventura - Palmira - Armenia": {
        "distancia": 230,
        "trafico": 3,
        "peajes": 2
    }
}

#F. Heurística
def heuristica(ruta):
    distancia = rutas[ruta]["distancia"]
    trafico = rutas[ruta]["trafico"]
    peajes = rutas[ruta]["peajes"]

    # Costo
    costo = distancia + (trafico * 2) + peajes
    return costo


#Mejor ruta
mejor_ruta = min(rutas, key=heuristica)

print("La mejor ruta para llegar al Parque del Café es:", mejor_ruta)
print("Fue la mejor ruta con un costo de:", heuristica(mejor_ruta))
```
