```python
#Rutas
rutas = {
    "Ruta Centro": {"distancia": 10, "trafico": 15},
    "Ruta Alterna": {"distancia": 14, "trafico": 5},
    "Ruta Malecon": {"distancia": 2, "trafico": 10}
}

# F. Heurística
def heuristica(ruta):
    distancia = rutas[ruta]["distancia"]
    trafico = rutas[ruta]["trafico"]

    # Costo
    costo_estimado = distancia + trafico
    return costo_estimado

# Mejor ruta
mejor_ruta = min(rutas, key=heuristica)

print("La mejor ruta para llegar al Puerto de Buneventura es:", mejor_ruta)
print("Fue la mejor ruta con un costo de:", heuristica(mejor_ruta))
```
