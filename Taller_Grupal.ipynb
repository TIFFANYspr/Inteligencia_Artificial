## <b> Crear el DataFrame </b>

Aqui creamos el DataFrame utilizando  Pandas y los datos proporcionados y pedidos en el taller y subidos al GitHub.

La función shape permite conocer la cantidad de filas y columnas dentro del DataFrame.

## <b> Generar una descripción estadística del conjunto de los datos </b>

Se utiliza describe() para obtener estadísticas generales del DataFrame.

# <b> PUNTO 2 — OPERACIONES MATEMÁTICAS </b>

En esta sección se crearan nuevas columnas utilizando operaciones matemáticas.

## <b> Crear columna Total_Venta </b>

Se multiplica el precio por la cantidad para calcular el total vendido por producto.  
Total_Venta = Precio × Cantidad

```python
df["Total_Venta"] = df["Precio"] * df["Cantidad"]

print(df[["Producto", "Total_Venta"]].head())
```

## <b> Crear columna Costo_Total </b>

Se suma el Total_Venta más el costo de envío.  

Costo_Total=Total_Venta+Costo_Envio

```python
df["Costo_Total"] = df["Total_Venta"] + df["Costo_Envio"]

print(df[["Producto", "Costo_Total"]].head())
```

## <b> Promedio de Total_Venta </b>

Se calcula el promedio de ventas utilizando mean().

```python
promedio_ventas = df["Total_Venta"].mean()

print("El promedio total de las ventas es:", promedio_ventas)
```

## <b> Producto con mayor Total_Venta </b>

Se identifica el producto con mayores ventas utilizando idxmax().

```python
produto_mayor_venta = df.loc[df["Total_Venta"].idxmax()]

print("El producto con mayor venta es:")
print(produto_mayor_venta)
```

## <b> Producto con menor Total_Venta </b>

Se identifica el producto con menores ventas utilizando idxmin().

```python
producto_menor_venta = df.loc[df["Total_Venta"].idxmin()]

print("El producto con menor venta es:")
print(producto_menor_venta)
```

# <b> PUNTO 3 — PORCENTAJES Y ANÁLISIS </b>

En esta sección se realizan cálculos de porcentajes relacionados con las ventas.


## <b>Crear columna IVA</b>

Se calcula el 19% del Total_Venta.

IVA = Total_Venta X 0.19

```python
df["IVA"] = df["Total_Venta"] * 0.19

print(df[["Producto", "IVA"]].head())
```

## <b> Crear columna Ganancia_Estimada </b>

Se calcula una ganancia estimada del 25%.  

Ganancia_Estimada = Total_Venta X 0.25

```python
df["Ganancia_Estimada"] = df["Total_Venta"] * 0.25

print(df[["Producto", "Ganancia_Estimada"]].head())
```

## <b> Crear columna Perdida_Estimada </b>

Se calcula una pérdida estimada del 0.05%.  

Perdida_Estimada = Total_Venta X 0.0005

```python
df["Perdida_Estimada"] = df["Total_Venta"] * 0.0005

print(df[["Producto", "Perdida_Estimada"]].head())
```

## <B>Calcular porcentaje de ventas</B>

Se calcula qué porcentaje representa cada producto respecto al total general.  

Porcentaje_Ventas = (Total_Venta/Total General)​ X 100

```python
total_general = df["Total_Venta"].sum()

df["Porcentaje_Ventas"] = (df["Total_Venta"] / total_general) * 100

print(df[["Producto", "Porcentaje_Ventas"]].head())
```

## <b>Filtrar productos con porcentaje superior al 15% </b>

Se muestran únicamente los productos cuyo porcentaje sea mayor al 15%.



```python
productos_mayor_15 = df[df["Porcentaje_Ventas"] > 15]

print(productos_mayor_15)
```

# <B> PUNTO 4 _ ESTADÍSTICA DESCRIPTIVA </B>

En esta sección se calculan diferentes medidas estadísticas para analizar el comportamiento de los datos.  

## <B> Calcular la media de la columna Precio </B>

La media representa el promedio de todos los precios registrados en el DataFrame.  

Media = ∑x ​/ n

```python
media = df["Precio"].mean()

print("La media del precio es:", media)
```

## <B> Calcular la mediana de la columna Precio </B>

La mediana representa el valor central de todos los precios ordenados.


```python
mediana = df["Precio"].median()

print("La mediana del precio es:", mediana)
```

## <B> Calcular la moda de la columna Precio </B>

La moda representa el valor que más se repite dentro de la columna Precio.

```python
moda = df["Precio"].mode()

print("La moda del precio es:", moda)

print(moda)
```

## <b> Calcular desviación estándar de Cantidad </b>

La desviación estándar mide qué tan dispersos están los datos respecto al promedio.  

σ = √(Σ(x - μ)² / N)


```python
desviacion = df["Cantidad"].std()

print("La desviación estándar es:", desviacion)
```

## <b>Calcular valor máximo y mínimo de Total_Venta</b>

Se identifica el valor más alto y el más bajo de las ventas totales.


```python
maximo = df["Total_Venta"].max()

minimo = df["Total_Venta"].min()

print("El valor máximo es:", maximo)

print("El valor mínimo es:", minimo)
```

## <b>Interpretación de Resultados</b>

- La media representa el promedio general de precios.
- La mediana representa el valor central de los precios.
- La moda representa el precio que más se repite.
- La desviación estándar indica qué tan dispersas están las cantidades.
- El valor máximo representa la venta más alta.
- El valor mínimo representa la venta más baja.

 # <B> PUNTO 5 - FILTRADO Y ORDENAMIENTO </B>

En esta sección se aplican filtros y ordenamientos sobre el DataFrame.  

## <B> Mostrar productos de Tecnología </B>

Se filtran únicamente los productos pertenecientes a la categoría Tecnología.



```python
productos_tecnologia = df[df["Categoria"] == "Tecnología"]

print(productos_tecnologia)
```
