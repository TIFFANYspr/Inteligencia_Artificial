```python
# Estrategias posibles
estrategias = ["Frenar para recoger pasajeros", "Seguir avanzando"]

# Matriz de pagos (Colectiva A, Colectiva B)
matriz = [
    [(4,4), (12,3)],   # A frena
    [(3,11), (7,8)]    # A sigue
]

print("Competencia entre dos colectivas por pasajeros\n")

print("Matriz de pagos (A,B):\n")

for i in range(2):
    for j in range(2):
        print(f"A: {estrategias[i]} | B: {estrategias[j]} -> {matriz[i][j]}")

print("\nAnálisis estratégico:\n")

print("Si B frena, A prefiere frenar (4 > 3)")
print("Si B sigue, A prefiere frenar (12 > 7)")

print("Si A frena, B prefiere frenar (4 > 3)")
print("Si A sigue, B prefiere frenar (11 > 8)")

print("\nEquilibrio encontrado:")
print("Ambas colectivas deciden frenar -> resultado (4,4)")
```
