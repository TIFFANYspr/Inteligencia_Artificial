```python

estrategias = ["Parqueadero cercano", "Parqueadero lejano"]

# (Conductor A, Conductor B)
matriz = [
    [(2,2), (9,4)],   # A intenta cercano
    [(4,9), (6,6)]    # A va al lejano
]

print("Juego de decisiones de parqueadero en el Edificio Torremar\n")

print("Matriz de pagos (A,B):\n")

for i in range(2):
    for j in range(2):
        print(f"A: {estrategias[i]} | B: {estrategias[j]} -> {matriz[i][j]}")

print("\nAnálisis de resultados:\n")

for i in range(2):
    for j in range(2):
        a, b = matriz[i][j]

        print(f"Decisiones:")
        print(f"Conductor A elige: {estrategias[i]}")
        print(f"Conductor B elige: {estrategias[j]}")
        print(f"Beneficio A = {a}")
        print(f"Beneficio B = {b}")
        print("-----------------------------")

print("En conclusión:")
print("Un existe un equilibrio cuando ambos eligen el parqueadero lejano (6,6)")
```
