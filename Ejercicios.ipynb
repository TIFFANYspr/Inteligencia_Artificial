```python
# =========================================
#         ACTIVIDAD PANDAS EN PYTHON
# =========================================

# Importar librerías
import pandas as pd
import numpy as np

# =========================================
#     EJERCICIO 1: CREAR UN DATAFRAME
# =========================================

print("========== RESULTADO EJERCICIO 1 ==========\n")

datos = {
    "Nombre": ["Juan", "María", "Pedro"],
    "Edad": [19, 22, 21],
    "Ciudad": ["Bogotá", "Cali", "Medellín"]
}

df1 = pd.DataFrame(datos)

print(df1)

# =========================================
#  EJERCICIO 2: OPERACIONES ENTRE COLUMNAS
# =========================================

print("\n========== RESULTADO EJERCICIO 2 ==========\n")

ventas = {
    "Producto": ["Zapatos", "Camisas", "Gorras"],
    "Precio": [120000, 80000, 30000],
    "Cantidad": [2, 3, 5]
}

df2 = pd.DataFrame(ventas)

df2["Total"] = df2["Precio"] * df2["Cantidad"]

print(df2)

# =========================================
#     EJERCICIO 3: LEER UN ARCHIVO CSV
# =========================================

print("\n========== RESULTADO EJERCICIO 3 ==========\n")

datos_csv = {
    "Producto": ["Camisa", "Pantalón", "Zapatos"],
    "Precio": [50000, 80000, 120000]
}

df_csv = pd.DataFrame(datos_csv)

# Crear archivo CSV
df_csv.to_csv("ventas.csv", index=False)

# Leer archivo CSV
df3 = pd.read_csv("ventas.csv")

print(df3.head())

# =========================================
#      EJERCICIO 4: DATOS SINTÉTICOS
# =========================================

print("\n========== RESULTADO EJERCICIO 4 ==========\n")

datos_sinteticos = {
    "Temperatura": np.random.randint(20, 35, 15),
    "Humedad": np.random.randint(40, 90, 15)
}

df4 = pd.DataFrame(datos_sinteticos)

print(df4)
```
