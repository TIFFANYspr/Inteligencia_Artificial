```python
Introducción

En este trabajo se realizó una investigación sobre la librería Pandas en Python y posteriormente se desarrollaron ejercicios prácticos para comprender cómo funciona el análisis y manipulación de datos.

El objetivo principal fue aprender a utilizar herramientas básicas de Pandas mediante ejemplos prácticos como la creación de DataFrames, operaciones matemáticas entre columnas, lectura de archivos CSV y generación de datos sintéticos.

A continuación, se presenta una explicación detallada de cada línea de código utilizada en la actividad y la función que cumple dentro del programa.

Objetivos
Objetivo general

Comprender el funcionamiento de Pandas mediante el análisis detallado de cada línea de código utilizada en los ejercicios prácticos.

Objetivos específicos
Comprender el propósito de cada línea de código.
Aprender cómo funcionan los DataFrames.
Analizar operaciones matemáticas entre columnas.
Entender el manejo de archivos CSV.
Comprender el uso de datos sintéticos con NumPy.
Importación de librerías
import pandas as pd

Esta línea importa la librería Pandas al programa. El alias pd permite escribir instrucciones más cortas y organizadas al momento de trabajar con DataFrames y análisis de datos.

import numpy as np

Aquí se importa la librería NumPy utilizando el alias np. Esta librería se usa principalmente para operaciones matemáticas y generación de números aleatorios.

EJERCICIO 1 — Crear un DataFrame
print("========== RESULTADO EJERCICIO 1 ==========")

La función print() muestra texto en pantalla para identificar el inicio del ejercicio y separar visualmente los resultados.

datos = {
    "Nombre": ["Juan", "María", "Pedro"],
    "Edad": [19, 22, 21],
    "Ciudad": ["Bogotá", "Cali", "Medellín"]
}

En esta parte se crea un diccionario llamado datos que almacena información organizada en columnas.

La columna "Nombre" contiene los nombres de las personas, "Edad" almacena sus edades y "Ciudad" guarda las ciudades correspondientes.

df1 = pd.DataFrame(datos)

Esta línea transforma el diccionario en un DataFrame de Pandas, creando una tabla organizada en filas y columnas similar a una hoja de Excel.

print(df1)

Muestra el DataFrame completo en pantalla para visualizar los datos organizados.

EJERCICIO 2 — Operaciones entre columnas
ventas = {
    "Producto": ["Zapatos", "Camisas", "Gorras"],
    "Precio": [120000, 80000, 30000],
    "Cantidad": [2, 3, 5]
}

Se crea un nuevo diccionario llamado ventas con información relacionada con productos, precios y cantidades vendidas.

df2 = pd.DataFrame(ventas)

Convierte el diccionario ventas en un DataFrame para trabajar la información de manera organizada.

df2["Total"] = df2["Precio"] * df2["Cantidad"]

Aquí se crea una nueva columna llamada "Total" multiplicando automáticamente el precio por la cantidad de cada producto.

Pandas realiza la operación fila por fila sin necesidad de hacer los cálculos manualmente.

print(df2)

Muestra el DataFrame actualizado incluyendo la nueva columna "Total".

EJERCICIO 3 — Leer archivos CSV
datos_csv = {
    "Producto": ["Camisa", "Pantalón", "Zapatos"],
    "Precio": [50000, 80000, 120000]
}

Se crea un diccionario con productos y precios que posteriormente será guardado en un archivo CSV.

df_csv = pd.DataFrame(datos_csv)

Convierte los datos en un DataFrame para poder manipularlos con Pandas.

df_csv.to_csv("ventas.csv", index=False)

Esta línea crea un archivo llamado ventas.csv y guarda dentro de él la información del DataFrame.

La opción index=False evita que Pandas agregue una columna extra con números de índice.

df3 = pd.read_csv("ventas.csv")

Lee el archivo CSV creado anteriormente y lo convierte nuevamente en un DataFrame de Pandas.

print(df3.head())

La función head() muestra las primeras filas del DataFrame para verificar que la información fue cargada correctamente.

EJERCICIO 4 — Datos sintéticos
datos_sinteticos = {
    "Temperatura": np.random.randint(20, 35, 15),
    "Humedad": np.random.randint(40, 90, 15)
}

En esta parte se generan datos aleatorios utilizando NumPy.

La función randint() crea números enteros aleatorios dentro de un rango específico.

En el caso de la temperatura, se generan 15 valores entre 20 y 35. Para la humedad, se generan 15 valores entre 40 y 90.

df4 = pd.DataFrame(datos_sinteticos)

Convierte los datos sintéticos en un DataFrame organizado para facilitar su análisis.

print(df4)

Muestra en pantalla la tabla con los datos aleatorios generados.
```
